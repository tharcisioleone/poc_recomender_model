```python
import json, sys, importlib

SETUP_CONFIG = "/Workspace/Users/tharcisio.leone@x1f-fink.one/poc_recomender_model/config/00_setup.json"
with open(SETUP_CONFIG) as f:
    setup = json.load(f)

sys.path.insert(0, setup["project_path"])

for mod in [m for m in list(sys.modules.keys()) if m.startswith("source")]:
    del sys.modules[mod]

EDA = importlib.import_module("source.03_eda").EDA

CONFIG_PATH = f"{setup['project_path']}/config/03_eda.json"
```

```python
eda = EDA(config_path=CONFIG_PATH)
reports = eda.run()
```

```python
for name, info in reports.items():
    print(f"\n=== {name} ===")
    print(f"  Rows: {info['rows']:,}")
    print(f"  Columns: {info['columns']}")
    print(f"  Report: {info['report_path']}")
    missing = {k: f"{v:.1%}" for k, v in info['missing_pct'].items() if v > 0}
    if missing:
        print(f"  Missing values: {missing}")
```
