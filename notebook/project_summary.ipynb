**POC — End-to-End Data Engineering on Databricks**

Kaggle Book-Crossing Dataset · Medallion Architecture · ALS Recommender

*Tharcisio Leone*

## Data

The [Kaggle Book-Crossing Dataset](https://www.kaggle.com/datasets/arashnic/book-recommendation-dataset) has 3 CSV files:

| File | Content |
|------|---------|
| `Books.csv` | ISBN, title, author, year of publication, publisher |
| `Users.csv` | User ID, location, age |
| `Ratings.csv` | User ID, ISBN, rating (0 = implicit, 1-10 = explicit) |

## Goal

Build a modular, reproducible pipeline on Databricks that turns the raw CSVs into clean, quality-checked tables, then uses them to:
* show reading trends (popular books, top authors, rating patterns) in Plotly, Lakeview and Power BI
* recommend books to each user with an ALS collaborative filtering model trained on the 378,030 explicit ratings

## Pipeline Architecture

| CSV | → | BRONZE | → | SILVER | → | GOLD | → | INSIGHTS |
|:---:|:---:|:---:|:---:|:---:|:---:|:---:|:---:|:---:|
| Ingestion | | Raw data | | Cleaned | | Enriched | | ML + Viz |
| `01_read_data` | | `bronze.*_raw` | | `04` + `06` DQ | | `05_data_merge` | | `08` + `09` + `10` + `11` |

**Quarantine (07):** records that fail the silver DQ rules are isolated in `quarantine.*`.

## Results

| Model | Params | Test RMSE | Test MAE |
|-------|--------|-----------|----------|
| 09a Baseline ALS | rank 10, regParam 0.1, maxIter 10 | 3.3060 | 2.5567 |
| 10 CV-tuned ALS | rank 100, regParam 0.5, maxIter 20 | 1.3305 | 0.8995 |

Cross-validation (18 combinations x 3 folds) cuts the test RMSE by about 60%. Predictions are clipped to the 1-10 rating scale.
