# 11 – ML Performance: Model Evaluation

Compares and visualizes the recommender models' performance in two sections.

**Section 1 — Model Comparison** (both models, from MLflow):
- Side-by-side RMSE and MAE between Step 09a (single-fit ALS) and Step 10 (CV-tuned ALS)

**Section 2 — Step 10 Results** (CV-tuned ALS only, from `gold.recommendations`):

- Top 25 most frequently recommended books
- Recommendation coverage (book and user)

> **Note:** The four charts in Section 2 reflect whichever model last wrote to `gold.recommendations`. Currently that is **Step 10 (CV-tuned ALS)**.

All charts saved into a single **HTML report**.

**Source:** `gold.{book_ratings, recommendations}`
**Output:** `reports/11_ml_performance_report.html`
**Config:** `config/11_ml_performance.json`

```python
import shutil, os

src = "/Workspace/Users/tharcisio.leone@x1f-fink.one/poc"
dst = "/Workspace/Users/tharcisio.leone@x1f-fink.one/poc_recomender_model"

for d in ["config", "source", "data", "files", "reports"]:
    src_dir = os.path.join(src, d)
    dst_dir = os.path.join(dst, d)
    if os.path.exists(src_dir):
        if os.path.exists(dst_dir):
            shutil.rmtree(dst_dir)
        shutil.copytree(src_dir, dst_dir)
        count = sum(len(f) for _, _, f in os.walk(dst_dir))
        print(f"  ✓ {d}/ ({count} files)")

if os.path.exists(os.path.join(src, ".gitignore")):
    shutil.copy2(os.path.join(src, ".gitignore"), os.path.join(dst, ".gitignore"))
    print("  ✓ .gitignore")

# Copy notebooks (they are .ipynb files in /notebook/)
src_nb = os.path.join(src, "notebook")
dst_nb = os.path.join(dst, "notebook")
if os.path.exists(src_nb):
    if os.path.exists(dst_nb):
        shutil.rmtree(dst_nb)
    shutil.copytree(src_nb, dst_nb)
    count = sum(len(f) for _, _, f in os.walk(dst_nb))
    print(f"  ✓ notebook/ ({count} files)")

print("\nAll files copied to Git folder.")
```
