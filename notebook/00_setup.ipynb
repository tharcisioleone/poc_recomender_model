```python
import json

CONFIG_PATH = "/Workspace/Users/tharcisio.leone@x1f-fink.one/poc_recomender_model/config/00_setup.json"
with open(CONFIG_PATH) as f:
    config = json.load(f)

for schema in config["schemas"]:
    spark.sql(f"CREATE DATABASE IF NOT EXISTS {schema}")
    print(f"Schema ready: {schema}")
```

```python
import os

for f in config["expected_files"]:
    full_path = f"{config['data_path']}/{f}"
    exists = os.path.exists(full_path)
    size_mb = os.path.getsize(full_path) / 1024 / 1024 if exists else 0
    status = f"OK ({size_mb:.1f} MB)" if exists else "MISSING"
    print(f"{f}: {status}")
```
