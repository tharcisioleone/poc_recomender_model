# 09b – Comparison Model: SVD++ via Surprise

## Approach

**SVD++** extends standard SVD by incorporating **implicit feedback** — it learns not only from rating values but also from *which* books a user chose to rate. This additional signal regularises latent factors and typically improves accuracy on explicit-rating datasets.

**Library:** [Surprise](https://surpriselib.com/) (`scikit-surprise`) — single-node, purpose-built for recommender research.

## ALS vs SVD++

| Aspect | ALS (09a) | SVD++ (09b) |
| --- | --- | --- |
| Library | PySpark ML | Surprise |
| Execution | Distributed | Single-node |
| Implicit feedback | No | Yes |
| Bias terms | No | Yes (user + item) |

## Pipeline

1. Load `gold.book_ratings` → convert to Surprise `Dataset`
2. Train/test split (80/20, seed 42)
3. Fit `SVDpp` with configurable `n_factors`, `n_epochs`, `lr_all`, `reg_all`
4. Evaluate RMSE / MAE on held-out test set
5. Log metrics to MLflow for comparison in step 11
