# 01 – Ingestion: CSV to Bronze

Reads the Kaggle Book-Crossing CSV files into bronze Delta tables.

**Source:** `data/kaggle/{Books,Users,Ratings}.csv`
**Target:** `bronze.books_raw`, `bronze.users_raw`, `bronze.ratings_raw`
**Config:** `config/01_read_data.json`

```
config/01_read_data.json    ← Paths, separator, table names
source/01_read_data.py      ← ReadData class (extract/transform/load)
notebook/01_read_data       ← This notebook (execution)
```

````python
import json, sys, importlib

SETUP_CONFIG = "/Workspace/Users/tharcisio.leone@x1f-fink.one/poc/config/00_setup.json"
with open(SETUP_CONFIG) as f:
    setup = json.load(f)

sys.path.insert(0, setup["project_path"])

for mod in [m for m in list(sys.modules.keys()) if m.startswith("source")]:
    del sys.modules[mod]

ReadData = importlib.import_module("source.01_read_data").ReadData

CONFIG_PATH = f"{setup['project_path']}/config/01_read_data.json"
````

````python
pipeline = ReadData(config_path=CONFIG_PATH)
dataframes = pipeline.run()
````
