# 02 – Data Governance

Applies governance policies to bronze tables:
- **Object-level:** Table and column comments (metadata documentation)
- **Tags:** Domain, layer, PII classification
- **Row-level security:** Filter policies (documented; requires Unity Catalog)
- **Column-level security:** Column masking for PII fields (documented; requires UC)

**Config:** `config/02_data_governance.json`

```python
import json, sys, importlib

SETUP_CONFIG = "/Workspace/Users/tharcisio.leone@x1f-fink.one/poc_recomender_model/config/00_setup.json"
with open(SETUP_CONFIG) as f:
    setup = json.load(f)

sys.path.insert(0, setup["project_path"])

for mod in [m for m in list(sys.modules.keys()) if m.startswith("source")]:
    del sys.modules[mod]

DataGovernance = importlib.import_module("source.02_data_governance").DataGovernance

CONFIG_PATH = f"{setup['project_path']}/config/02_data_governance.json"
```

```python
governance = DataGovernance(config_path=CONFIG_PATH)
results = governance.run()
```

```python
print("=== Governance Summary ===")
for item in results["summary"]:
    print(f"  {item['table']}: {item['columns']} columns documented")

print("\n=== Security Policies (documented for UC migration) ===")
for mask_sql in results["security_policies"]["column_masking"]:
    print(mask_sql)
for filter_sql in results["security_policies"]["row_filters"]:
    print(filter_sql)
```
