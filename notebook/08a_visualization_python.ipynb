```python
import json, sys, importlib

SETUP_CONFIG = "/Workspace/Users/tharcisio.leone@x1f-fink.one/poc/config/00_setup.json"
with open(SETUP_CONFIG) as f:
    setup = json.load(f)

sys.path.insert(0, setup["project_path"])

for mod in [m for m in list(sys.modules.keys()) if m.startswith("source")]:
    del sys.modules[mod]

DataVisualization = importlib.import_module("source.08_data_visualization").DataVisualization

CONFIG_PATH = f"{setup['project_path']}/config/08_data_visualization.json"
```

```python
viz = DataVisualization(config_path=CONFIG_PATH)
results = viz.run()

print(f"HTML report saved: {results['report_path']}")
print(f"Charts generated: {len(results['figures'])}")
```

```python
print("=" * 60)
print("Chart parameters (from config/08_data_visualization.json)")
print("=" * 60)
d = viz._defaults
print(f"  top_n              : {viz.chart_cfg.get('top_n', d['top_n'])}")
print(f"  min_ratings        : {viz.chart_cfg.get('min_ratings', d['min_ratings'])}")
print(f"  min_ratings_per_group: {viz.chart_cfg.get('min_ratings_per_group', d['min_ratings_per_group'])}")
print(f"  title_max_chars    : {viz.chart_cfg.get('title_max_chars', d['title_max_chars'])}")
print(f"  min_year (resolved): {viz._min_year}")
print(f"  max_year (resolved): {viz._max_year}")
manual = "min_year_manual" in viz.chart_cfg
print(f"  year range source  : {'manual override' if manual else 'dataset min/max'}")
print("=" * 60)
print()

for fig in results["figures"]:
    fig.show()
```
